```r
library(tidyverse)

employees <- data.frame(
  Name = c("Ahmed", "Ali", "Omar", "Sara", "Mona"),
  Age = c(25, 30, 28, 24, 32),
  Department = c(
    "IT", "HR", "IT", "Finance", "HR"
  ),
  Salary = c(5000, 6000, 7000, 4500, 8000)
)

employees
```

```r
# First: Selecting specific columns using `select`

employees %>%
  select(Name, Salary)
```

```r
# Second: Filtering data using `filter`

employees %>%
  filter(Salary > 5000)
```

```r
# Third: Sorting salaries using `arrange`

employees %>%
  arrange(desc(Salary))
```

```r
# Fourth: Calculating the average salary using `summarise`

employees %>%
  summarise(
    Average_Salary = mean(Salary),
    Maximum_Salary = max(Salary),
    Minimum_Salary = min(Salary)
  )
```

```r
# Fifth: Grouping data by category using `group_by`

employees %>%
  group_by(Department) %>%
  summarise(
    Average_Salary = mean(Salary),
    Employees_Count = n()
  )
```
