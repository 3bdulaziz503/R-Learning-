```r
# 4. Variables
# A variable is a name we use to store a value that we can retrieve later.
# In R, we often use the symbol <- to assign values.

name <- "Ahmed"

age <- 25

salary <- 5000
```

```r
# To display the values:

print(name)
print(age)
print(salary)
```

```r
# We can display more than one value:

print(name)
print(age)
print(salary)
```

```r
# 5. Data Types
# There are several basic data types in R.
# First: Numeric – decimal numbers

price <- 150.75
```

```r
# Second: Integer – Whole Numbers

quantity <- 10L
```

```r
# Third: Character – Texts

country <- "Saudi Arabia"
```

```r
# Fourth: Logical – Logical values

is_student <- TRUE
is_employee <- FALSE
```

```r
# Fifth: Complex – Complex Numbers

z <- 3 + 2i
```

```r
# How do we determine the type of a variable?
# We use the class() function:

price <- 150.75

class(price)
```

```r
# Try the other varieties:
class(quantity)
class(country)
class(is_student)
class(z)
```

```r
# 6. Arithmetic Operations
# R supports standard arithmetic operations:
# Try running the following program:

a <- 20
b <- 6

addition <- a + b
subtraction <- a - b
multiplication <- a * b
division <- a / b
power <- a ^ 2
remainder <- a %% b

print(addition)
print(subtraction)
print(multiplication)
print(division)
print(power)
print(remainder)

```

```r
# 7. Retrieving Information About Variables
# There are some important functions that you will use frequently:
# class()	معرفة نوع الكائن
# typeof()	معرفة نوع التخزين الداخلي
# length()	معرفة عدد العناصر
# str()	    عرض بنية الكائن 
# print()	طباعة القيمة
# ls()	    عرض أسماء المتغيرات الموجودة   

student <- "Ahmed"

class(student)

typeof(student)

length(student)

str(student)
```
