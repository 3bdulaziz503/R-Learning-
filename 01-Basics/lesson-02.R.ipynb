```r
grades <- c(85, 90, 75, 95, 80)

print(grades)
```

# 3. Creating a Character Vector

We can store a collection of names within a single vector:

```r
names <- c("Ahmed","Mohammed","Ali","Omar")

print(names)
```

Note that text in R must be enclosed in quotation marks.

# 4. Creating a Logical Vector

A logical vector contains two fundamental values:

•	TRUE, which means true.

•	FALSE, which means false.

```r
results <- c(TRUE, FALSE, TRUE, TRUE, FALSE)

print(results)
```
